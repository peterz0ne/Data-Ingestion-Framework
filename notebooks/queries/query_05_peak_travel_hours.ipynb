# Query 5: Peak Travel Hours by Weekday

Taxi demand is defined as the number of trips beginning in each weekday and local pickup-hour combination.

## Connect to Spark

Start the Docker stack from this repository before running these cells. The query runner reads the Gold Delta table and registers it as the temporary SQL view `integrated_taxi_trips`.

```python
import sys

from pyspark.sql import SparkSession

JOB_PATH = '/opt/spark/jobs'
if JOB_PATH not in sys.path:
    sys.path.insert(0, JOB_PATH)

from analytics.runner import execute_query

spark = (
    SparkSession.builder
    .master('spark://spark-master:7077')
    .appName('Assignment2-Query5-PeakTravelHours')
    .getOrCreate()
)
spark.conf.set('spark.sql.session.timeZone', 'UTC')
spark
```

## Execute the analytical query

The result shows the highest-demand pickup hour for every weekday. `pickup_hour_local` is New York local time. If multiple hours tie for the highest demand on a weekday, all tied peak hours are returned.

```python
query_5_result = execute_query(
    spark,
    'query_05_peak_travel_hours',
)

query_5_result.show(truncate=False)
```
