# Query 3: Air Quality and Taxi Demand

Taxi demand is defined as the number of taxi trips beginning in each pickup hour.

```python
import sys

from pyspark.sql import SparkSession

JOB_PATH = '/opt/spark/jobs'
if JOB_PATH not in sys.path:
    sys.path.insert(0, JOB_PATH)

from analytics.runner import execute_query

spark = (
    SparkSession.builder
    .master('spark://spark-master:7077')
    .appName('Assignment2-Query3-AirQualityDemand')
    .getOrCreate()
)
spark.conf.set('spark.sql.session.timeZone', 'UTC')
spark
```
