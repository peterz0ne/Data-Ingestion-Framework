```python
import pandas as pd
```

```python
taxi_zones = pd.read_csv("taxi_zone_lookup.csv")

# basic structure and inferred data types
print(f"Shape: {taxi_zones.shape}\n")
taxi_zones.info()

# summary of each column
column_summary = pd.DataFrame({
    "dtype": taxi_zones.dtypes.astype(str),
    "missing": taxi_zones.isna().sum(),
    "unique_values": taxi_zones.nunique(dropna=False),
    "suggested_type": [
        "category" if dtype == "object" and unique / len(taxi_zones) < 0.5 # if the unique-total ratio is less than 50%, consider it a category
        else str(dtype)
        for dtype, unique in zip(taxi_zones.dtypes, taxi_zones.nunique())
    ]
})

display(column_summary)

# frequency tables for each column
for column in taxi_zones.columns:
    print(f"\n--- {column} ---")
    display(taxi_zones[column].value_counts(dropna=False).head(20))

# preview the data
display(taxi_zones.head())
```
