```python
from pyspark.sql import SparkSession

spark = (
    SparkSession.builder
    .master("spark://spark-master:7077")
    .appName("MyNotebook")
    .getOrCreate()
)

spark
```

```python
gold = spark.read.format("delta").load("s3a://delta-data/silver/gold_integrated_taxi_trips")

```

```python
gold.createOrReplaceTempView("gold")
```

```python

 spark.sql("""
    select count(1),cast(pickup_datetime as date)
    from gold 
    group by cast(pickup_datetime as date)
""").show()

```
